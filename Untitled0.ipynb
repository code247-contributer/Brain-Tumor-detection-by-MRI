```python
import tensorflow as tf
from tensorflow.keras import layers, models

DATA = "/content/data"
IMG = (224, 224)

def load(path, **kw):
    ds = tf.keras.utils.image_dataset_from_directory(
        path, image_size=IMG, batch_size=32,
        label_mode="int", **kw)
    names = ds.class_names
    ds = ds.map(lambda x, y: (tf.cast(x, tf.float32), tf.cast(y, tf.int32)))
    return ds.prefetch(tf.data.AUTOTUNE), names

train, classes = load(DATA + "/Training", validation_split=0.2,
                      subset="training", seed=42)
val, _ = load(DATA + "/Training", validation_split=0.2,
              subset="validation", seed=42)
test, _ = load(DATA + "/Testing")
print(classes)

base = tf.keras.applications.MobileNetV2(
    input_shape=IMG + (3,), include_top=False, weights="imagenet")
base.trainable = False

model = models.Sequential([
    layers.Input(shape=IMG + (3,)),
    layers.Rescaling(1./127.5, offset=-1),
    base,
    layers.GlobalAveragePooling2D(),
    layers.Dropout(0.3),
    layers.Dense(4, activation="softmax")])

model.compile("adam", "sparse_categorical_crossentropy", ["accuracy"])
model.fit(train, validation_data=val, epochs=5)
print(model.evaluate(test))
```

```python
import numpy as np, os
import tensorflow as tf
from tensorflow.keras import layers, models
from tensorflow.keras.utils import load_img, img_to_array

DATA = "/content/data"
IMG = (224, 224)
classes = sorted(os.listdir(DATA + "/Training"))
print(classes)

def load_folder(path, limit=None):
    X, y = [], []
    for i, c in enumerate(classes):
        files = sorted(os.listdir(f"{path}/{c}"))[:limit]
        for f in files:
            img = load_img(f"{path}/{c}/{f}", target_size=IMG)
            X.append(img_to_array(img))
            y.append(i)
    return np.array(X, dtype="float32"), np.array(y, dtype="int32")

# limit=800 se loading jaldi hoti hai (~2-3 min); poora data chahiye to limit=None kar do
X, y = load_folder(DATA + "/Training", limit=800)
Xt, yt = load_folder(DATA + "/Testing", limit=300)
print(X.shape, Xt.shape)

idx = np.random.RandomState(42).permutation(len(X))
X, y = X[idx], y[idx]

base = tf.keras.applications.MobileNetV2(
    input_shape=IMG + (3,), include_top=False, weights="imagenet")
base.trainable = False

model = models.Sequential([
    layers.Input(shape=IMG + (3,)),
    layers.Rescaling(1./127.5, offset=-1),
    base,
    layers.GlobalAveragePooling2D(),
    layers.Dropout(0.3),
    layers.Dense(4, activation="softmax")])

model.compile(optimizer="adam", loss="sparse_categorical_crossentropy", metrics=["accuracy"])
model.fit(X, y, validation_split=0.2, epochs=5, batch_size=32)
print(model.evaluate(Xt, yt))
```

```python
import matplotlib.pyplot as plt

ids = np.random.RandomState(1).choice(len(Xt), 8, replace=False)
preds = model.predict(Xt[ids]).argmax(axis=1)

plt.figure(figsize=(12, 7))
for k, i in enumerate(ids):
    plt.subplot(2, 4, k + 1)
    plt.imshow(Xt[i].astype("uint8"))
    plt.title(f"True: {classes[yt[i]]}\nPred: {classes[preds[k]]}", fontsize=9)
    plt.axis("off")
plt.show()
```
